```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Neha'],
    'Marks': [85, np.nan, 92, np.nan]
}

df = pd.DataFrame(data)

print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum().sum())
```

```python
print(df.notnull())
```

```python
print(df.dropna())
```

```python
data = {
    'Name': ['Aman', 'Riya', np.nan],
    'Marks': [85, 90, np.nan]
}

df = pd.DataFrame(data)

print(df.dropna(how='all'))
```

```python
print(df.fillna(0))
```

```python
data = {
    'Name': ['Aman', 'Riya', 'Rahul'],
    'Marks': [85, np.nan, 90]
}

df = pd.DataFrame(data)

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print(df)
```

```python
data = {
    'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday'],
    'Sales': [100, np.nan, 150, np.nan]
}

df = pd.DataFrame(data)

print(df['Sales'].ffill())
```

```python
print(df['Sales'].bfill())
```

```python

data = {
    'Student': ['Aman', 'Riya', 'Rahul', 'Neha'],
    'Marks': [85, np.nan, 92, np.nan],
    'Attendance': [90, 95, np.nan, 88]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nAfter Filling Missing Values:")
print(df.fillna(0))

print("\nAfter Dropping Rows with Missing Values:")
print(df.dropna())
```

```python
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    'Age': [20, 25, np.nan, 30, 35]
})

imputer = SimpleImputer(strategy='mean')

df['Age'] = imputer.fit_transform(df[['Age']])

print(df)
```

```python
df = pd.DataFrame({
    'Salary': [30000, 35000, np.nan, 40000, 100000]
})

imputer = SimpleImputer(strategy='median')

df['Salary'] = imputer.fit_transform(df[['Salary']])

print(df)
```

```python
df = pd.DataFrame({
    'Department': ['IT', 'HR', np.nan, 'IT', 'Finance']
})

imputer = SimpleImputer(strategy='most_frequent')

df['Department'] = imputer.fit_transform(df[['Department']]).ravel()

print(df)
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df = pd.DataFrame({
    'Hours_Studied': [5, 6, np.nan, 8, 9],
    'Marks': [60, 65, 70, np.nan, 85]
})

imputer = IterativeImputer(max_iter=10, random_state=42)

df[['Hours_Studied', 'Marks']] = imputer.fit_transform(
    df[['Hours_Studied', 'Marks']]
)

print(df)
```

```python
df = pd.DataFrame({
    'Age': [20, 25, np.nan, 35, 40],
    'Salary': [30000, 35000, 40000, np.nan, 100000],
    'Department': ['IT', 'HR', np.nan, 'IT', 'Finance']
})

mean_imputer = SimpleImputer(strategy='mean')
df['Age_Mean'] = mean_imputer.fit_transform(df[['Age']])

median_imputer = SimpleImputer(strategy='median')
df['Salary_Median'] = median_imputer.fit_transform(df[['Salary']])

mode_imputer = SimpleImputer(strategy='most_frequent')
df['Department_Mode'] = mode_imputer.fit_transform(
    df[['Department']]
).ravel()

mice_imputer = IterativeImputer(max_iter=10, random_state=42)

df[['Age_MICE', 'Salary_MICE']] = mice_imputer.fit_transform(
    df[['Age', 'Salary']]
)

print(df)
```
