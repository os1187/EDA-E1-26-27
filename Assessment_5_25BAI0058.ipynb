```python
df = pd.read_csv("WineQT.csv")

print(df.dtypes)
```

```python
print(df.describe())
```

```python
 mean = df["alcohol"].mean()
median = df["alcohol"].median()
mode = df["alcohol"].mode()

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
print("Mean:", df["quality"].mean())
print("Median:", df["quality"].median())
print("Mode:", df["quality"].mode())
```

```python
print("Standard Deviation of Alcohol:",
      df["alcohol"].std())

print("Standard Deviation of Quality:",
      df["quality"].std())
```

```python
print("Variance of Alcohol:",
      df["alcohol"].var())

print("Variance of Quality:",
      df["quality"].var())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["alcohol"], kde=True)

plt.title("Distribution of Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.histplot(df["quality"], kde=True)

plt.title("Distribution of Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.boxplot(x="alcohol", data=df)

plt.title("Box Plot of Alcohol")
plt.show()
```

```python
sns.boxplot(x="quality", y="alcohol", data=df)

plt.title("Alcohol Distribution by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol")

plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])

plt.title("Alcohol vs Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")

plt.show()
```

```python
plt.scatter(df["alcohol"], df["sulphates"])

plt.title("Alcohol vs Sulphates")
plt.xlabel("Alcohol")
plt.ylabel("Sulphates")

plt.show()
```

```python
sns.regplot(x="alcohol", y="quality", data=df)

plt.title("Regression Plot: Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")

plt.show()
```

```python
sns.pairplot(
    df,
    vars=["alcohol", "pH", "sulphates", "quality"]
)

plt.show()
```

```python
sns.pairplot(
    df,
    vars=["alcohol", "pH", "sulphates", "quality"],
    kind="reg"
)

plt.show()
```

```python
from scipy import stats
corr = stats.pearsonr(
    df["alcohol"],
    df["quality"]
)

print("p-value:", corr[1])
print("Correlation:", corr[0])
```

```python
corr = stats.pearsonr(
    df["fixed acidity"],
    df["citric acid"]
)

print("p-value:", corr[1])
print("Correlation:", corr[0])
```

```python
correlation = df.corr(
    method="pearson",
    numeric_only=True
)

print(correlation)
```

```python
correlation = df.corr(
    method="pearson",
    numeric_only=True
)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Wine Dataset")

plt.show()
```

```python
correlation = df.corr(numeric_only=True)

sns.heatmap(correlation)

plt.title("Wine Dataset Correlation Heatmap")

plt.show()
```
